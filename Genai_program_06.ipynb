```python
#6.Use a pre-trained Hugging Face model to analyze sentiment in text. Assume a real-world application,
#Load the sentiment analysis pipeline. Analyze the sentiment by giving sentences to input.
```

```python
!pip install transformers torch

from transformers import pipeline

sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

reviews = [
    "The product is absolutely fantastic! I love it.",
    "I'm not satisfied with the quality of this product.",
    "Great customer service, would definitely recommend!",
    "This is the worst purchase I've ever made. Totally regret it.",
    "I feel neutral about the product, it's okay, not great but not bad either."
]

for review in reviews:
    result = sentiment_analyzer(review)
    sentiment = result[0]["label"]
    confidence = result[0]["score"]

    print(f"Review: {review}")
    print(f"Sentiment: {sentiment} (Confidence: {confidence:.4f})")
    print()
```
