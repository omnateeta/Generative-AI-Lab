```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)
```

```python
long_text = """
The process of machine learning involves training algorithms on data to allow
them to make predictions or decisions based on new data.

It begins with gathering large sets of data and then using this data to train
models. The models learn patterns and relationships
within the data, which can be used to make informed predictions when new data is
presented.

Machine learning is used in various fields including healthcare, finance, and
technology, where it aids in tasks such as medical diagnosis, fraud detection,
and customer service automation.

Over time, machine learning models continue to improve as they are exposed to
more data, becoming increasingly accurate and effective at their tasks.
"""
```

```python
summary = summarizer(
    long_text,
    max_length=100,
    min_length=30,
    do_sample=False
)
```

```python
print("Original Text:\n")
print(long_text)

print("\nSummarized Text:\n")
print(summary[0]["summary_text"])
```
